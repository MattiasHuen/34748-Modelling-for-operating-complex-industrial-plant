# **System Decomposition - fundamental concepts to approach the complexity | Notes**

When facing bigger problems, we break them down to smaller problems. Pretty straightforward thinking wise. There exists multiple ways, depending on the problem.

## **Automation System**
Complexity in engineered systems require system decomposition. This includes some sort of hierarchy and distribution in industrial automation. By distribution, we mean physical strucres and temporal distribution. By Hierarchies, we mean ontologies, control/command hierarchy, goal hierarchy and part-whole hierarchy.

**Levels of control**

0. Primary Technology.
1. Group control, Unit control, Field, Sensors and actors.
2. Supervisory. (SCADA = Supervisory control and data acquisition)
3. WOrkflow, order tracking, resources.
4. Production planning, orders, purchases.
5. Planning, Statistics, Finances.



### **Hierarchy**
The formal definition: A set of nodes in a tree strucutre, (directed) with relation of subordination. Each node belongs to a level in the hierarchy, and a node can only be subordinate to one other node. The relation is therefor transitive (No A is subordinate to C, you can essentially collaps a subordinate relationsship) and can have any number of levels.

As possoed to heterarchies, where a node can be subordinate to more than one other node, the hierarchy tree structure can be thought of a one way ticket upwards.

### **Distribution**
Distribution in **space**, is about control systems and process (Generation, Consumption). Distribution in time, is about sequence of actions, parallel processing and manufacturing steps, essentially includes some sort of timeline (production-line).

## **System Decomposition in 2 Dimensions**
A complex system can be discribed on two different dimensions.

- Means-end decomposition
- Part whole decomposition

Operation of a system is based on the system's goals/objectives. Function is the behaviour that supports the goals, and are realized by components that can expose multiple behaviours.
