## **HAZOP**

Hazop is required by regulation (Law)

Hazop stands for Hazard and operability analysis. The analysis is made to identify possible hazards in the operation. The HAZOP manual isn't sutied for the operator at the console, since its very long and text based, meaning a stressfull situation it would be hard to use since its a document (difficult to search). If a hazard does occur, it might not be related to the situations in the Hazard analysis, meaning from a console person standpoint, it would be overall hard to use since they experience situations that arent described in the documents.
