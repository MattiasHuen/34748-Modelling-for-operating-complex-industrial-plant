**Ruled-based system**

A Rule-based system contrains the following components:
- A list of rules or a rule base.
- An inference engine or semantic reasoner.
    - Inference engine task cycle:
        - Match
        - Resolve conflicts
        - act
    - A working memory (fact base)
    - A user interface (read in new facts and present reasoning conclusion)
- Commercial rule engine: Drools etc.
